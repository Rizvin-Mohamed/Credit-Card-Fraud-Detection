```python
import pandas as pd

X_test = pd.read_csv("../data/X_test.csv")
y_test = pd.read_csv("../data/y_test.csv")["Class"]
test = X_test.assign(Class=y_test)

demo = pd.concat([test[test["Class"] == 1],
                  test[test["Class"] == 0].sample(400, random_state=42)])
demo = demo.sample(frac=1, random_state=42).reset_index(drop=True)
demo.to_csv("../app/demo_transactions.csv", index=False)

sample = pd.read_csv("../data/sample_statement.csv")
sample[["datetime", "merchant", "amount", "city"]].to_csv("../app/sample_statement.csv", index=False)

print("Demo transactions:", demo.shape, "| frauds:", demo["Class"].sum())
print("Sample statement:", sample.shape[0], "rows")
```
